```python
#6 Array Stacking and Spiltting 
```

```python
a = np.array([1,2,3])
b = np.array([4,5,6])

print(np.vstack((a,b)))  # vertical stacking -row wise
print(np.hstack((a,b))) # horixontal stacking - column wise
```

```python
c = np.array([[1,2,3], [4,5,6]])
print(c)
```

```python
hsplit = np.hsplit(c,3) # horizontal split

for s in hsplit:
    print(s)
```

```python
vsplit = np.vsplit(c,2) # vartical split

for s in vsplit:
    print(s)
```
